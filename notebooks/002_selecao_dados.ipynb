## Criando os dataframes

```python
import pandas as pd

jogos = pd.read_csv(r"../data/bruto/games.csv")
jogadores = pd.read_csv(r"../data/bruto/players.csv")
valores_jogadores = pd.read_csv(r"../data/bruto/player_valuations.csv")
clubes = pd.read_csv(r"../data/bruto/clubs.csv")
```

## Verificando código do Brasileirão

```python
jogos["competition_id"].unique()
```

## Filtrando pelos jogos do Brasileirao

```python
import os

os.makedirs("../data/processado", exist_ok=True)

jogos_br = jogos[jogos["competition_id"] == "BRA1"].copy()

cobertura = jogos_br.groupby("season")["date"].agg(["min", "max", "count"])
jogos_br.to_csv("../data/processado/jogos_brasileirao.csv", index=False)
cobertura
```
